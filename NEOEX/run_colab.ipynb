Este notebook instala e executa a versão 1.10. Envie o arquivo `NEOEX_v1.10.zip` na primeira célula.

> Se você já executou uma versão anterior, reinicie primeiro o ambiente de execução do Colab para eliminar módulos e servidores antigos.


```python
from google.colab import files
from pathlib import Path
import shutil, zipfile

uploaded = files.upload()  # selecione NEOEX_v1.10.zip
zip_name = next((name for name in uploaded if name.lower().endswith('.zip')), None)
if zip_name is None:
    raise RuntimeError('Envie o arquivo ZIP do projeto NEOEX v1.10.')

project_dir = Path('/content/NEOEX')
if project_dir.exists():
    shutil.rmtree(project_dir)
with zipfile.ZipFile(zip_name) as archive:
    bad = archive.testzip()
    if bad is not None:
        raise RuntimeError(f'Arquivo ZIP corrompido: {bad}')
    roots = {Path(name).parts[0] for name in archive.namelist() if Path(name).parts}
    if 'NEOEX' not in roots:
        raise RuntimeError('O ZIP precisa conter a pasta NEOEX na raiz.')
    archive.extractall('/content')
if not (project_dir / 'app.py').exists():
    raise RuntimeError('Não encontrei app.py dentro da pasta NEOEX.')
print('NEOEX v1.10 extraído com sucesso.')
print('Arquivos:', ', '.join(sorted(p.name for p in project_dir.iterdir())))
```
